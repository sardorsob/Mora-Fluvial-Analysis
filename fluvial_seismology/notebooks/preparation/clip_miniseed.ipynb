# Extract time windows from miniSEED recordings

Upstream research notebook from `taylor-kenyon/FluvialSeismology_Analysis`, `clip_mseed.ipynb`, commit `b4962248fdcfed970c0f5c64091472ab1933f2fc`. Saved outputs were cleared. Review inputs and any local paths before running.

```python
from obspy import read, UTCDateTime
import os

input_path = r"D:\calibration_rename\NOC2511\GPZ\ZE.C2511..GPZ_20251023215340.mseed"

st = read(input_path)

# check time range
print(st)
print(st[0].stats.starttime, st[0].stats.endtime)

# Set the clipping window for this example.
#start = UTCDateTime("2026-08-18T12:15:00")
start = UTCDateTime(2025, 10, 23, 22, 3, 15)
end = UTCDateTime(2025, 10, 23, 22, 7, 5) # ~4 mins
#end = start + 2 * 60  # 2 minutes later

# Trim a copy so the in-memory stream retains the original samples.
st_clip = st.copy().trim(starttime=start, endtime=end)

# Write the clip to a separate output folder.
output_dir = r"D:\clipped_mseed"
os.makedirs(output_dir, exist_ok=True)

output_filename = "ZE.C2511..GPZ_20251023215340_clipped.mseed"
output_path = os.path.join(output_dir, output_filename)

# extra safety: refuse to write if a file already exists at that path
if os.path.exists(output_path):
    raise FileExistsError(f"{output_path} already exists — rename or delete it first.")

st_clip.write(output_path, format="MSEED")
print(f"Saved clipped file to: {output_path}")
```
